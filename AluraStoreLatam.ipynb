### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
tienda2.head()
```

#1. Análisis de facturación



```python
facturacion_total = {
    "Tienda 1": tienda["Precio"].sum(),
    "Tienda 2": tienda2["Precio"].sum(),
    "Tienda 3": tienda3["Precio"].sum(),
    "Tienda 4": tienda4["Precio"].sum()
}

facturacion_total
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(facturacion_total.keys(), facturacion_total.values(), color='skyblue')
plt.title("Facturación Total por Tienda")
plt.ylabel("Total en pesos")
plt.xlabel("Tienda")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Análisis de Facturación
En este apartado evaluamos la facturación total de cada tienda.

- Tienda 1 tiene la mayor facturación.
- Tienda 4 es la que menos vendió.


# 2. Ventas por categoría

```python
# Tienda 1 - Ventas por categoría
ventas_categoria_1 = tienda.groupby("Categoría del Producto")["Precio"].sum().sort_values(ascending=False)

# Tienda 2
ventas_categoria_2 = tienda2.groupby("Categoría del Producto")["Precio"].sum().sort_values(ascending=False)

# Tienda 3
ventas_categoria_3 = tienda3.groupby("Categoría del Producto")["Precio"].sum().sort_values(ascending=False)

# Tienda 4
ventas_categoria_4 = tienda4.groupby("Categoría del Producto")["Precio"].sum().sort_values(ascending=False)
ventas_categoria_1
```

```python
ventas_categoria_2
```

```python
ventas_categoria_3
```

```python
ventas_categoria_4
```

**Mi analisis es que la tendencia en las tiendas es igual para todas, lo que cambian son los volumenes y estan relacionadas con la facturacion total, es decir tienda 1 vende mas y tienda 4 vende menos, las ventas por categoria seria algo redundante y no me aporta algo adicional de lo que ya me aporta el punto 1 de facturación total**

```python
plt.figure(figsize=(10, 6))
ventas_categoria_1.plot(kind='bar', color='salmon')
plt.title("Ventas por Categoría - Tienda 1")
plt.ylabel("Total en pesos")
plt.xlabel("Categoría del Producto")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# 3. Calificación promedio de la tienda


```python

calificacion_promedio = {
    "Tienda 1": tienda["Calificación"].mean(),
    "Tienda 2": tienda2["Calificación"].mean(),
    "Tienda 3": tienda3["Calificación"].mean(),
    "Tienda 4": tienda4["Calificación"].mean(),
}

calificacion_promedio
```

**La tienda 1 tiene la menor nota y la tienda 3 la mayor nota, aunque en terminos númericos todas las tiendas están alrededor de 4 de calificación**

```python
plt.figure(figsize=(8, 5))
plt.bar(calificacion_promedio.keys(), calificacion_promedio.values(), color='lightgreen')
plt.title("Calificación Promedio por Tienda")
plt.ylabel("Promedio de Calificación")
plt.xlabel("Tienda")
plt.ylim(0, 5)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# 4. Productos más y menos vendidos

```python
# Productos más y menos vendidos - Tienda 1
productos_t1 = tienda["Producto"].value_counts()
print("Tienda 1 - Más vendidos:\n", productos_t1.head(5))
print("\nTienda 1 - Menos vendidos:\n", productos_t1.tail(5))

# Productos más y menos vendidos - Tienda 2
productos_t2 = tienda2["Producto"].value_counts()
print("\nTienda 2 - Más vendidos:\n", productos_t2.head(5))
print("\nTienda 2 - Menos vendidos:\n", productos_t2.tail(5))

# Productos más y menos vendidos - Tienda 3
productos_t3 = tienda3["Producto"].value_counts()
print("\nTienda 3 - Más vendidos:\n", productos_t3.head(5))
print("\nTienda 3 - Menos vendidos:\n", productos_t3.tail(5))

# Productos más y menos vendidos - Tienda 4
productos_t4 = tienda4["Producto"].value_counts()
print("\nTienda 4 - Más vendidos:\n", productos_t4.head(5))
print("\nTienda 4 - Menos vendidos:\n", productos_t4.tail(5))

```

Cuando realizamos un análisis de los productos más o menos vendidos hay una tendencia similar con cambios mínimos en los detalles por lo que consideramos que esta variable no da una tendencia

# 5. Envío promedio por tienda

```python
envio_promedio = {
    "Tienda 1": tienda["Costo de envío"].mean(),
    "Tienda 2": tienda2["Costo de envío"].mean(),
    "Tienda 3": tienda3["Costo de envío"].mean(),
    "Tienda 4": tienda4["Costo de envío"].mean()
}

for tienda, costo in envio_promedio.items():
    print(f"{tienda}: ${costo:,.2f}")
```
